# Cell delimitation on wax strips
Cellpose (GPU) segments every cell; only filled (brown) cells are kept. Run in the `wax-cells` kernel.

```python
import matplotlib.pyplot as plt
import cell_detection as cd

PDF, PAGE = '20261109145816_SecureScan.pdf', 2   # page index 0-3
img = cd.load_page(PDF, page=PAGE, dpi=200)      # BGR image
labels = cd.segment(img)          # filled cells: [(cx, cy, r)], label image
print(labels.max(), 'filled cells')
```

```python
out = cd.draw_contours(img, labels)
plt.figure(figsize=(10, 14)); plt.imshow(out[..., ::-1]); plt.axis('off'); plt.show()
```
